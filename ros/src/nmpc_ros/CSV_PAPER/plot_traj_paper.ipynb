```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.lines import Line2D
import seaborn as sns
from PIL import Image
import matplotlib.image as mpimg
from matplotlib.offsetbox import OffsetImage, AnnotationBbox
from matplotlib.ticker import MultipleLocator
import matplotlib.patches as patches
from matplotlib.legend_handler import HandlerBase
import io

# ==========================================
# 1. SETUP PALETTE E COLORI (Dal Notebook originale)
# ==========================================
try:
    logo_arr = mpimg.imread("/home/pantheon/drea/neural_mpc/ros/src/nmpc_ros/results/sun_orientation.png")
except:
    print("Logo file not found, creating a placeholder.")
    logo_arr = np.zeros((10, 10, 4)) 

paired_palette = sns.color_palette("Paired")
dark_paired_palette = sns.color_palette("dark")

# Colormap per le auto (Verde, Grigio, Rosso) in base all'esito (lambda)
custom_cmap = LinearSegmentedColormap.from_list(
    "custom_cmap", [paired_palette[3], (0.5, 0.5, 0.5), dark_paired_palette[1]]
)

class DummyGradient: pass

# ==========================================
# 2. PARAMETRI GEOMETRICI (Dal file plot_vincoli)
# ==========================================
safe_radius_pole = 0.6
poles_pos = np.array([
    [6.5905, 6.178], [9.7331, 5.1122], [16.6093, 4.0214],
    [20.3781, 3.3552], [26.0805, 2.1324], [29.9115, 1.6864],
    [35.5549, 0.3543], [39.2697, -0.2548], [45.0900, -1.2590],
    [48.4675, -1.9691], [53.8411, -3.1006], [56.1188, -3.4193],
], dtype=np.float32)

margin_f = 0.5
flowerbeds_pos = np.array([
    [2.1879, -5.5748, 4.6071, 4.5434, -0.175],
    [64.6609, -16.5047, 2.8938, 4.5661, -0.175],
    [32.1876, -15.5858, 66.7007, 3.5897, -0.175],
    [3.8231, 4.0245, 1.6749, 4.7252, -0.175],
    [5.7205, 7.9094, 3.5549, 3.5383, -0.175],
    [61.5571, -4.8938, 1.2004, 7.0567, -0.175],
    [9.7648, 16.3899, 2.0120, 4.6903, -0.175],
    [58.9274, 7.9500, 2.4176, 4.3564, -0.175],
    [34.1108, 15.1653, 50.4020, 1.5882, -0.175],
], dtype=np.float32)

car_length = 3.46 
car_width = 1.62   
margin_x = 0.6    
margin_y = 0.5 
core_safety = 0.8 

# ==========================================
# 3. HANDLERS LEGGENDA E FUNZIONI DI SUPPORTO
# ==========================================
class HandlerGradient(HandlerBase):
    def __init__(self, cmap, num_stripes=64, **kw):
        super().__init__(**kw)
        self.cmap = cmap
        self.num_stripes = num_stripes

    def create_artists(self, legend, orig_handle, xdescent, ydescent, width, height, fontsize, trans):
        gradient = np.linspace(0, 1, self.num_stripes)
        colors = self.cmap(gradient)
        stripes = []
        for i in range(self.num_stripes):
            stripe = patches.Rectangle([xdescent + i * width / self.num_stripes, ydescent],
                                     width / self.num_stripes, height,
                                     fc=colors[i], transform=trans, edgecolor='none')
            stripes.append(stripe)
        return stripes

class HandlerImage(HandlerBase):
    def __init__(self, image_arr, zoom=1.0, **kw):
        super().__init__(**kw)
        self.image_arr = image_arr
        self.zoom = zoom

    def create_artists(self, legend, orig_handle, xdescent, ydescent, width, height, fontsize, trans):
        imagebox = OffsetImage(self.image_arr, zoom=self.zoom)
        ab = AnnotationBbox(imagebox, (xdescent + 0.5 * width, ydescent + 0.5 * height),
                            frameon=False, xycoords=trans, box_alignment=(0.5, 0.5), pad=0)
        return [ab]

def generate_superellipse(center_x, center_y, a, b, theta, n=4, num_points=200):
    """Genera i punti (x, y) di una superellisse ruotata."""
    t = np.linspace(0, 2 * np.pi, num_points)
    power = 2.0 / n
    x_rel = a * np.sign(np.cos(t)) * (np.abs(np.cos(t)) ** power)
    y_rel = b * np.sign(np.sin(t)) * (np.abs(np.sin(t)) ** power)
    dx = x_rel * np.cos(theta) - y_rel * np.sin(theta)
    dy = x_rel * np.sin(theta) + y_rel * np.cos(theta)
    return center_x + dx, center_y + dy

def get_latest_csv(mode, directory, suffix="plot_data.csv"):
    pattern = os.path.join(directory, f"{mode}*{suffix}")
    files = glob.glob(pattern)
    if not files:
        return None
    return max(files, key=os.path.getmtime)

def load_csv_data(csv_path):
    with open(csv_path, 'r') as f:
        first_line = f.readline().strip()
    parts = first_line.split(',')
    if parts[0] == "tree_positions":
        tree_positions_list = [float(x) for x in parts[1:] if x]
        if tree_positions_list:
            num_trees = len(tree_positions_list) // 3
            tree_positions = np.array(tree_positions_list).reshape(num_trees, 3)
        else:
            tree_positions = np.array([]).reshape(0,3)
    else:
        tree_positions = None
    df = pd.read_csv(csv_path, skiprows=2)
    return (
        df["time"].values, df["x"].values, df["y"].values, df["theta"].values,
        df["entropy"].values, df[[col for col in df.columns if col.startswith("lambda_")]].values,
        tree_positions
    )

# ==========================================
# 4. FUNZIONI DI PLOTTING MIGLIORATE
# ==========================================
def plot_trajectory_subplot(ax, x, y, theta, trees, lambda_history, custom_cmap,
                            tree_label_fontsize, legend_fontsize_param,
                            axis_label_fontsize, tick_label_fontsize,
                            subplot_title_fontsize, mode_label_for_title, 
                            obstacle_core_color, obstacle_core_edge, obstacle_margin_color, 
                            obstacle_margin_edge, excluded_car_ids=None):
    
    # Plot Traiettoria Drone
    ax.plot(x, y, marker='o', color='orange', linewidth=2.5, markersize=1.5, label="Drone Trajectory")
    ax.scatter(x[0], y[0], color='crimson', s=250, marker='X', label="Initial Position", zorder=6)
    ax.scatter(x[-1], y[-1], color='gold', s=250, marker='*', label="Final Position", zorder=6)

    # ==========================================
    # VINCOLI HARD - PALI
    # ==========================================
    for p in poles_pos:
        # 1. Margine di Sicurezza
        margin_circle = patches.Circle(
            (p[0], p[1]), safe_radius_pole, 
            facecolor=obstacle_margin_color, edgecolor=obstacle_margin_edge, 
            linewidth=0.8, zorder=1
        )
        ax.add_patch(margin_circle)
        
        # 2. Struttura Fisica Solida del Palo
        pole_core = patches.Circle(
            (p[0], p[1]), 0.25, 
            facecolor=obstacle_core_color, edgecolor=obstacle_core_edge, 
            linewidth=0.8, zorder=3
        )
        ax.add_patch(pole_core)

    # ==========================================
    # VINCOLI HARD - AIUOLE
    # ==========================================
    for f in flowerbeds_pos:
        fx, fy, flen, fwid, ftheta = f[0], f[1], f[2], f[3], f[4]
        
        # --- A. MARGINE DI SICUREZZA ---
        sigma_x_f = (flen / 2.0) + margin_f
        sigma_y_f = (fwid / 2.0) + margin_f
        x_se_margin, y_se_margin = generate_superellipse(fx, fy, sigma_x_f, sigma_y_f, ftheta)
        
        ax.fill(
            x_se_margin, y_se_margin, 
            facecolor=obstacle_margin_color, edgecolor=obstacle_margin_edge, 
            linewidth=1.0, zorder=1
        )
        
        # --- B. AIUOLA REALE ---
        sigma_x_core = flen / 2.0
        sigma_y_core = fwid / 2.0
        x_se_core, y_se_core = generate_superellipse(fx, fy, sigma_x_core, sigma_y_core, ftheta)
        
        ax.fill(
            x_se_core, y_se_core, 
            facecolor=obstacle_core_color, edgecolor=obstacle_core_edge, 
            linewidth=1.2, zorder=2
        )

    # Plot Auto/Macchine (Trees nel tuo CSV) con Superellissi
    if trees is not None and trees.size > 0:
        for i in range(trees.shape[0]):

            # if excluded_car_ids is not None and i in excluded_car_ids:
            #     continue
            # final_lambda = lambda_history[-1, i*2]
            # tree_color = custom_cmap(final_lambda)

            if excluded_car_ids is not None and i in excluded_car_ids:
                tree_color = 'white'
            else:
                final_lambda = lambda_history[-1, i*2]
                tree_color = custom_cmap(final_lambda)
            
            car_x = trees[i, 0]
            car_y = trees[i, 1]
            car_theta = trees[i, 2]
            
            # Traslazione al centro geometrico
            center_x = car_x - (car_length / 2.0) * np.cos(car_theta)
            center_y = car_y - (car_length / 2.0) * np.sin(car_theta)
            
            sigma_x = (car_length / 2.0) + margin_x
            sigma_y = (car_width / 2.0) + margin_y
            a_hard = sigma_x * core_safety
            b_hard = sigma_y * core_safety
            
            # Disegna la superellisse dell'auto con il colore basato sull'entropia
            x_se, y_se = generate_superellipse(center_x, center_y, a_hard, b_hard, car_theta)
            ax.fill(x_se, y_se, color=tree_color, alpha=0.7, edgecolor='black', zorder=3)
            
            # Disegna l'orientamento dell'auto
            arrow_length = 1.5 
            dx = arrow_length * np.cos(car_theta)
            dy = arrow_length * np.sin(car_theta)
            ax.arrow(center_x, center_y, dx, dy, color='black', width=0.05, head_width=0.4, head_length=0.4, zorder=4)

    # Frecce Direzione Drone
    step = max(1, len(x) // 300)
    for idx_arr in range(0, len(x), step):
        x0, y0, t = x[idx_arr], y[idx_arr], theta[idx_arr]
        x1 = x0 + 1.5 * np.cos(t)
        y1 = y0 + 1.5 * np.sin(t)
        ax.annotate("", xy=(x1, y1), xytext=(x0, y0), arrowprops=dict(arrowstyle="->", color="orange", linewidth=1.5))

    ax.tick_params(axis='both', labelsize=tick_label_fontsize)
    ax.set_aspect('equal', adjustable='box')
    ax.set_xlim(-2, 68)
    ax.set_ylim(-24, 22)
    ax.set_title(mode_label_for_title, fontsize=subplot_title_fontsize)
    ax.xaxis.set_major_locator(MultipleLocator(5))
    ax.yaxis.set_major_locator(MultipleLocator(5))

def plot_entropy_subplot(ax, time_history, entropy, label, color, axis_label_fontsize, tick_label_fontsize):
    ax.plot(time_history, entropy, linewidth=1.5, markersize=1.0, label=label, color=color)
    ax.set_xlabel("Time (s)", fontsize=axis_label_fontsize)
    ax.set_ylabel("Entropy", fontsize=axis_label_fontsize)
    ax.tick_params(axis='both', labelsize=tick_label_fontsize)

# ==========================================
# 5. ESECUZIONE (Multi-mode Comparison)
# ==========================================
modes = ["nmpc", "mower"] 
baselines_dir = "./to_plot"

num_non_visitabili = 11

mower_excluded_car_ids = [4, 16, 22, 29]  # contiene anche non visitabili
mower_excluded_car_ids_visible =  [4, 16] # contiene solo visitabili

# valore minimo di entropia teoricamente raggiungibile
min_theoretical_entropy = 12 
entropy_offset_per_car = 1.0   # Valore da sottrarre per ciascuna auto non presente

axis_label_fontsize = 15
tick_label_fontsize = 16
subplot_title_fontsize = 22
legend_fontsize = 14
suptitle_fontsize = 18
tree_label_fontsize = 10

algorithm_labels = {
    "linear": "Linear Path",
    "greedy": "Greedy Approach",
    "mower": "Mower Path",
    "mower_good": "Mower Path - Good View",
    "mower_bad": "Mower Path - Poor View",
    "nmpc": "Neural IPP MPC",
}

# Colori uniformi per TUTTI gli ostacoli ambientali
obstacle_core_color = '#D4A373'      # Colore solido interno (Pali e Aiuole)
obstacle_core_edge = '#B08050'       # Bordo interno
obstacle_margin_color = '#FAEDCD'    # Margine di sicurezza esterno
obstacle_margin_edge = '#E9D8A6'    # Bordo del margine

nrows = 1
ncols = (len(modes) + nrows - 1) // nrows 

fig_traj, axs_traj = plt.subplots(nrows, ncols, figsize=(14, 6 * nrows), squeeze=False)
axs_flat = axs_traj.flatten() 
fig_entropy, ax_entropy = plt.subplots(figsize=(7, 3))

colors = sns.color_palette("tab10", len(modes))
any_trees = False

for i, mode in enumerate(modes):
    label = algorithm_labels.get(mode, mode.capitalize())
    csv_file = get_latest_csv(mode, baselines_dir)
    if not csv_file:
        print(f"No CSV found for {mode}")
        if i < len(axs_flat):
            axs_flat[i].axis('off')
        continue
    try:
        time_h, x, y, theta, entropy, lambda_h, trees = load_csv_data(csv_file)
        if trees is not None and trees.size > 0:
            any_trees = True
    except Exception as e:
        print(f"Error loading {csv_file}: {e}")
        if i < len(axs_flat):
            axs_flat[i].axis('off')
        continue

    # Seleziona gli ID da escludere solo per la modalità mower
    excluded_ids = mower_excluded_car_ids if mode == "mower" else None

    ax_t = axs_flat[i] 
    plot_trajectory_subplot(ax_t, x, y, theta, trees, lambda_h, custom_cmap,
                            tree_label_fontsize, legend_fontsize,
                            axis_label_fontsize, tick_label_fontsize,
                            subplot_title_fontsize, label, obstacle_core_color, obstacle_core_edge, 
                            obstacle_margin_color, obstacle_margin_edge,
                            excluded_car_ids=excluded_ids)
    oim = OffsetImage(logo_arr, zoom=0.075)
    ab = AnnotationBbox(oim, (-0.01, 1.01), xycoords="axes fraction", frameon=False)
    ax_t.add_artist(ab)

    entropy_to_plot = entropy.copy()
    if mode == "mower" and mower_excluded_car_ids_visible:
        total_offset = len(mower_excluded_car_ids_visible) * entropy_offset_per_car
        entropy_to_plot = entropy_to_plot - total_offset

    # Abbassa entropia totale per auto non visitabili
    entropy_to_plot = entropy_to_plot - num_non_visitabili

    plot_entropy_subplot(ax_entropy, time_h, entropy_to_plot, label, colors[i],
                         axis_label_fontsize, tick_label_fontsize)
    ax_entropy.axvline(x=time_h[-1], color=colors[i], linestyle='--', linewidth=1.5)

for i in range(len(modes), len(axs_flat)):
    axs_flat[i].axis('off')

# Finalize Trajectory Plot
legend_handles = [
    Line2D([0], [0], color='orange', lw=1.5, marker='o', markersize=2, label='Drone Trajectory'),
    Line2D([0], [0], marker='X', color='w', markerfacecolor='crimson', markersize=15, label='Initial Position'),
    Line2D([0], [0], marker='*', color='w', markerfacecolor='gold', markersize=25, label='Final Position'),
    patches.Patch(facecolor=obstacle_core_color, edgecolor=obstacle_core_edge, label='Obstacle'),
    patches.Patch(facecolor='white', edgecolor='black', linewidth=1.0, label='Removed Car')
]
legend_labels = ['Trajectory', 'Initial Position', 'Final Position', 'Obstacle', 'Removed Car']

# image_handle = patches.Patch(color='none')
# legend_handles.append(image_handle)

handler_map = {
    # image_handle: HandlerImage(logo_arr, zoom=0.04) 
}

if any_trees:
    gradient_handle = patches.Patch(color='none')
    legend_handles.append(gradient_handle)
    legend_labels.append('Empty(Green)-Full(Red) Belief')
    handler_map[gradient_handle] = HandlerGradient(cmap=custom_cmap)

fig_traj.tight_layout(rect=[0, 0.05, 1, 0.95])
fig_traj.legend(
    handles=legend_handles, 
    labels=legend_labels,
    loc='lower center', 
    bbox_to_anchor=(0.5, 0.0), 
    ncol=6, 
    fontsize=legend_fontsize,
    handler_map=handler_map
)

# ax_entropy.axhline(
#     y=min_theoretical_entropy, 
#     color='red', 
#     linestyle=':', 
#     linewidth=1.8, 
#     label='Min Reachable Entropy',
#     zorder=5
# )

output_path_traj = os.path.join(baselines_dir, f"thesis_mpc_trajectory_grid.png")
fig_traj.savefig(output_path_traj, format='png', bbox_inches='tight',  dpi=350)

output_path_traj_svg = os.path.join(baselines_dir, "thesis_mpc_trajectory_grid.svg")
fig_traj.savefig(output_path_traj_svg, format='svg', bbox_inches='tight')

# Finalize Entropy Plot
ax_entropy.set_xlim(0, 1285) 
ax_entropy.set_ylim(0, 22)
ax_entropy.set_title("Entropy Comparison", fontsize=suptitle_fontsize)
ax_entropy.legend(loc='lower left', bbox_to_anchor=(0.0, 0.05))
ax_entropy.grid(True, linestyle='--', alpha=0.9)
output_path_entropy = os.path.join(baselines_dir, f"entropy_comparison.png")
fig_entropy.savefig(output_path_entropy, format='png', bbox_inches='tight', dpi=350)
output_path_entropy_svg = os.path.join(baselines_dir, "entropy_comparison.svg")
fig_entropy.savefig(output_path_entropy_svg, format='svg', bbox_inches='tight')

plt.show()
```

```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.lines import Line2D
import seaborn as sns
from PIL import Image
import matplotlib.image as mpimg
from matplotlib.offsetbox import OffsetImage, AnnotationBbox
from matplotlib.ticker import MultipleLocator
import matplotlib.patches as patches
import matplotlib.animation as animation
from matplotlib.animation import FFMpegWriter
from matplotlib.legend_handler import HandlerBase
import io

# ==========================================
# 1. SETUP PALETTE E COLORI
# ==========================================
try:
    logo_arr = mpimg.imread("/home/pantheon/drea/neural_mpc/ros/src/nmpc_ros/results/sun_orientation.png")
except:
    print("Logo file not found, creating a placeholder.")
    logo_arr = np.zeros((10, 10, 4)) 

paired_palette = sns.color_palette("Paired")
dark_paired_palette = sns.color_palette("dark")

# Colormap per le auto (Verde, Grigio, Rosso) in base all'esito (lambda)
custom_cmap = LinearSegmentedColormap.from_list(
    "custom_cmap", [paired_palette[3], (0.5, 0.5, 0.5), dark_paired_palette[1]]
)

class DummyGradient: pass

# ==========================================
# 2. PARAMETRI GEOMETRICI
# ==========================================
safe_radius_pole = 0.6
poles_pos = np.array([
    [6.5905, 6.178], [9.7331, 5.1122], [16.6093, 4.0214],
    [20.3781, 3.3552], [26.0805, 2.1324], [29.9115, 1.6864],
    [35.5549, 0.3543], [39.2697, -0.2548], [45.0900, -1.2590],
    [48.4675, -1.9691], [53.8411, -3.1006], [56.1188, -3.4193],
], dtype=np.float32)

margin_f = 0.5
flowerbeds_pos = np.array([
    [2.1879, -5.5748, 4.6071, 4.5434, -0.175],
    [64.6609, -16.5047, 2.8938, 4.5661, -0.175],
    [32.1876, -15.5858, 66.7007, 3.5897, -0.175],
    [3.8231, 4.0245, 1.6749, 4.7252, -0.175],
    [5.7205, 7.9094, 3.5549, 3.5383, -0.175],
    [61.5571, -4.8938, 1.2004, 7.0567, -0.175],
    [9.7648, 16.3899, 2.0120, 4.6903, -0.175],
    [58.9274, 7.9500, 2.4176, 4.3564, -0.175],
    [34.1108, 15.1653, 50.4020, 1.5882, -0.175],
], dtype=np.float32)

car_length = 3.46 
car_width = 1.62   
margin_x = 0.6    
margin_y = 0.5 
core_safety = 0.8 

# ==========================================
# 3. HANDLERS LEGGENDA E FUNZIONI DI SUPPORTO
# ==========================================
class HandlerGradient(HandlerBase):
    def __init__(self, cmap, num_stripes=64, **kw):
        super().__init__(**kw)
        self.cmap = cmap
        self.num_stripes = num_stripes

    def create_artists(self, legend, orig_handle, xdescent, ydescent, width, height, fontsize, trans):
        gradient = np.linspace(0, 1, self.num_stripes)
        colors = self.cmap(gradient)
        stripes = []
        for i in range(self.num_stripes):
            stripe = patches.Rectangle([xdescent + i * width / self.num_stripes, ydescent],
                                     width / self.num_stripes, height,
                                     fc=colors[i], transform=trans, edgecolor='none')
            stripes.append(stripe)
        return stripes

class HandlerImage(HandlerBase):
    def __init__(self, image_arr, zoom=1.0, **kw):
        super().__init__(**kw)
        self.image_arr = image_arr
        self.zoom = zoom

    def create_artists(self, legend, orig_handle, xdescent, ydescent, width, height, fontsize, trans):
        imagebox = OffsetImage(self.image_arr, zoom=self.zoom)
        ab = AnnotationBbox(imagebox, (xdescent + 0.5 * width, ydescent + 0.5 * height),
                            frameon=False, xycoords=trans, box_alignment=(0.5, 0.5), pad=0)
        return [ab]

def generate_superellipse(center_x, center_y, a, b, theta, n=4, num_points=200):
    """Genera i punti (x, y) di una superellisse ruotata."""
    t = np.linspace(0, 2 * np.pi, num_points)
    power = 2.0 / n
    x_rel = a * np.sign(np.cos(t)) * (np.abs(np.cos(t)) ** power)
    y_rel = b * np.sign(np.sin(t)) * (np.abs(np.sin(t)) ** power)
    dx = x_rel * np.cos(theta) - y_rel * np.sin(theta)
    dy = x_rel * np.sin(theta) + y_rel * np.cos(theta)
    return center_x + dx, center_y + dy

def get_latest_csv(mode, directory, suffix="plot_data.csv"):
    pattern = os.path.join(directory, f"{mode}*{suffix}")
    files = glob.glob(pattern)
    if not files:
        return None
    return max(files, key=os.path.getmtime)

def load_csv_data(csv_path):
    with open(csv_path, 'r') as f:
        first_line = f.readline().strip()
    parts = first_line.split(',')
    if parts[0] == "tree_positions":
        tree_positions_list = [float(x) for x in parts[1:] if x]
        if tree_positions_list:
            num_trees = len(tree_positions_list) // 3
            tree_positions = np.array(tree_positions_list).reshape(num_trees, 3)
        else:
            tree_positions = np.array([]).reshape(0,3)
    else:
        tree_positions = None
    df = pd.read_csv(csv_path, skiprows=2)
    return (
        df["time"].values, df["x"].values, df["y"].values, df["theta"].values,
        df["entropy"].values, df[[col for col in df.columns if col.startswith("lambda_")]].values,
        tree_positions
    )

# ==========================================
# 4. FUNZIONE PER DISEGNARE GLI OSTACOLI FISSI
# ==========================================
def draw_static_environment(ax, mode_label_for_title, subplot_title_fontsize, tick_label_fontsize,
                             obstacle_core_color, obstacle_core_edge, obstacle_margin_color, obstacle_margin_edge):
    """Disegna solo i vincoli hard fissi (pali e aiuole) e la formattazione degli assi."""
    # Pali
    for p in poles_pos:
        margin_circle = patches.Circle((p[0], p[1]), safe_radius_pole, facecolor=obstacle_margin_color, edgecolor=obstacle_margin_edge, linewidth=0.8, zorder=1)
        ax.add_patch(margin_circle)
        pole_core = patches.Circle((p[0], p[1]), 0.25, facecolor=obstacle_core_color, edgecolor=obstacle_core_edge, linewidth=0.8, zorder=3)
        ax.add_patch(pole_core)

    # Aiuole
    for f in flowerbeds_pos:
        fx, fy, flen, fwid, ftheta = f[0], f[1], f[2], f[3], f[4]
        sigma_x_f, sigma_y_f = (flen / 2.0) + margin_f, (fwid / 2.0) + margin_f
        x_se_margin, y_se_margin = generate_superellipse(fx, fy, sigma_x_f, sigma_y_f, ftheta)
        ax.fill(x_se_margin, y_se_margin, facecolor=obstacle_margin_color, edgecolor=obstacle_margin_edge, linewidth=1.0, zorder=1)
        
        sigma_x_core, sigma_y_core = flen / 2.0, fwid / 2.0
        x_se_core, y_se_core = generate_superellipse(fx, fy, sigma_x_core, sigma_y_core, ftheta)
        ax.fill(x_se_core, y_se_core, facecolor=obstacle_core_color, edgecolor=obstacle_core_edge, linewidth=1.2, zorder=2)

    ax.tick_params(axis='both', labelsize=tick_label_fontsize)
    ax.set_aspect('equal', adjustable='box')
    ax.set_xlim(-2, 68)
    ax.set_ylim(-24, 22)
    ax.set_title(mode_label_for_title, fontsize=subplot_title_fontsize)
    ax.xaxis.set_major_locator(MultipleLocator(5))
    ax.yaxis.set_major_locator(MultipleLocator(5))

# ==========================================
# 5. SETUP E GENERAZIONE ANIMAZIONE MP4
# ==========================================
baselines_dir = "./to_plot"
mode_to_animate = "nmpc"  # Scegli 'nmpc' o 'mower'

algorithm_labels = {
    "linear": "Linear Path", "greedy": "Greedy Approach",
    "mower": "Mower Path", "nmpc": "Neural IPP MPC",
}
label = algorithm_labels.get(mode_to_animate, mode_to_animate.capitalize())

mower_excluded_car_ids = [4, 16, 22, 29]
excluded_ids = mower_excluded_car_ids if mode_to_animate == "mower" else []

obstacle_core_color, obstacle_core_edge = '#D4A373', '#B08050'
obstacle_margin_color, obstacle_margin_edge = '#FAEDCD', '#E9D8A6'
tick_label_fontsize, subplot_title_fontsize = 16, 22

csv_file = get_latest_csv(mode_to_animate, baselines_dir)
if csv_file is None:
    raise FileNotFoundError(f"Nessun file CSV trovato per la modalità: {mode_to_animate}")

time_h, x, y, theta, entropy, lambda_h, trees = load_csv_data(csv_file)

# Creazione Figura Animazione
fig_anim, ax_anim = plt.subplots(figsize=(12, 6))

# Disegna gli elementi di sfondo statici
draw_static_environment(
    ax_anim, f"Animation: {label}", subplot_title_fontsize, tick_label_fontsize,
    obstacle_core_color, obstacle_core_edge, obstacle_margin_color, obstacle_margin_edge
)

# Aggiunge logo in alto a sinistra
oim = OffsetImage(logo_arr, zoom=0.075)
ab = AnnotationBbox(oim, (-0.01, 1.01), xycoords="axes fraction", frameon=False)
ax_anim.add_artist(ab)

# Plot dei punti di Inizio e Fine
ax_anim.scatter(x[0], y[0], color='crimson', s=250, marker='X', label="Initial Position", zorder=6)
ax_anim.scatter(x[-1], y[-1], color='gold', s=250, marker='*', label="Final Position", zorder=6)

# ==========================================
# INIZIALIZZAZIONE ELEMENI DINAMICI AUTO (LAMBDA IN TEMPO REALE)
# ==========================================
car_patches = []

if trees is not None and trees.size > 0:
    for i in range(trees.shape[0]):
        car_x, car_y, car_theta = trees[i, 0], trees[i, 1], trees[i, 2]
        center_x = car_x - (car_length / 2.0) * np.cos(car_theta)
        center_y = car_y - (car_length / 2.0) * np.sin(car_theta)
        
        sigma_x = (car_length / 2.0) + margin_x
        sigma_y = (car_width / 2.0) + margin_y
        a_hard, b_hard = sigma_x * core_safety, sigma_y * core_safety
        
        x_se, y_se = generate_superellipse(center_x, center_y, a_hard, b_hard, car_theta)
        
        # Gestione auto escluse (es. per modalità mower)
        if excluded_ids is not None and i in excluded_ids:
            initial_color = 'white'
        else:
            initial_lambda = lambda_h[0, i * 2]
            initial_color = custom_cmap(initial_lambda)
        
        # Disegna la Patch della superellisse dell'auto
        poly_patch = patches.Polygon(np.column_stack([x_se, y_se]), facecolor=initial_color, alpha=0.7, edgecolor='black', zorder=3)
        ax_anim.add_patch(poly_patch)
        car_patches.append(poly_patch)
        
        # Freccia di orientamento dell'auto (statica)
        arrow_length = 1.5 
        dx, dy = arrow_length * np.cos(car_theta), arrow_length * np.sin(car_theta)
        ax_anim.arrow(center_x, center_y, dx, dy, color='black', width=0.05, head_width=0.4, head_length=0.4, zorder=4)

# Traiettoria e freccia drone dinamica
line_traj, = ax_anim.plot([], [], color='orange', linewidth=2.5, label="Drone Path")
drone_point, = ax_anim.plot([], [], marker='o', color='crimson', markersize=8, zorder=7)
drone_arrow = None

def init():
    line_traj.set_data([], [])
    drone_point.set_data([], [])
    return [line_traj, drone_point] + car_patches

# ==========================================
# FUNZIONE DI AGGIORNAMENTO AD OGNI FRAME
# ==========================================
def update(frame):
    global drone_arrow

    # 1. Aggiorna il percorso del drone
    line_traj.set_data(x[:frame+1], y[:frame+1])
    drone_point.set_data([x[frame]], [y[frame]])
    
    # 2. Aggiorna la freccia di orientamento del drone
    if drone_arrow is not None:
        drone_arrow.remove()
        
    x0, y0, t = x[frame], y[frame], theta[frame]
    x1, y1 = x0 + 1.8 * np.cos(t), y0 + 1.8 * np.sin(t)
    drone_arrow = ax_anim.annotate(
        "", xy=(x1, y1), xytext=(x0, y0),
        arrowprops=dict(arrowstyle="->", color="red", linewidth=2.0),
        zorder=7
    )

    # 3. AGGIORNAMENTO DINAMICO DEI COLORI DELLE AUTO IN BASE A LAMBDA(t)
    if trees is not None and trees.size > 0:
        for i in range(trees.shape[0]):
            if excluded_ids is not None and i in excluded_ids:
                continue
            
            current_lambda = lambda_h[frame, i * 2]
            current_color = custom_cmap(current_lambda)
            car_patches[i].set_facecolor(current_color)

    return [line_traj, drone_point, drone_arrow] + car_patches

# ==========================================
# SINCRONIZZAZIONE TEMPORALE E SALVATAGGIO
# ==========================================
subsample = 1  
frames = list(range(0, len(x), subsample))
if frames[-1] != len(x) - 1:
    frames.append(len(x) - 1)

total_real_time = time_h[-1] - time_h[0] 
total_frames = len(frames)
target_fps = total_frames / total_real_time

anim = animation.FuncAnimation(
    fig_anim, update, frames=frames, init_func=init, blit=False
)

plt.close(fig_anim)

mp4_output_path = os.path.join(baselines_dir, f"trajectory_{mode_to_animate}_synced.mp4")

print(f"--- Dati Sincronizzazione Animazione ---")
print(f"Durata simulazione reale: {total_real_time:.2f} s ({total_real_time/60:.2f} min)")
print(f"Numero totale di Frame: {total_frames}")
print(f"FPS calcolati per il montaggio: {target_fps:.3f}")
print(f"Generazione del video MP4 in corso...")

writer = FFMpegWriter(fps=target_fps, metadata=dict(artist='Drone Sim'), bitrate=3000)
anim.save(mp4_output_path, writer=writer, dpi=200)

print(f"\nVideo MP4 salvato con successo in:\n{mp4_output_path}")
```
