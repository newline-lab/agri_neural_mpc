```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Imposta lo stile per figure piccole ma leggibili
sns.set_theme(context="poster", style="whitegrid")
sns.set_context("poster")  # testi grandi

# Parametri
alpha_s = 10
d_th = 9.0

# Range di valori
d_min = np.linspace(0, 10, 400)
d_min_q = np.linspace(0, 6, 400)

# Funzioni
Q_d = 1e-2 / (1 + np.exp(alpha_s * (-d_min**2 + d_th)))
Q_d_q = 1 / (1 + np.exp(alpha_s * (-d_min_q**2 + d_th)))
C_a = Q_d * d_min**2

# Crea la figura
fig, axs = plt.subplots(1, 2, figsize=(8, 1.75))

# Plot Q_d
sns.lineplot(x=d_min_q, y=Q_d_q, ax=axs[0], color="royalblue", linewidth=2.0)
axs[0].set_xlabel(r'$d_{\min}$', labelpad=-3)
axs[0].set_ylabel(r'$Q_d$', labelpad=-15)
axs[0].grid(True, linewidth=0.3)

# Plot C_a
sns.lineplot(x=d_min, y=C_a, ax=axs[1], color="orange", linewidth=2.0)
axs[1].set_xlabel(r'$d_{\min}$', labelpad=-3)
axs[1].set_ylabel(r'$\mathcal{C}_a$', labelpad=-15)
axs[1].grid(True, linewidth=0.3)

# Salva
output_path = "../to_plot/activation_function.png"
plt.savefig(output_path, format='png', dpi=600, bbox_inches='tight')
plt.close()
```
